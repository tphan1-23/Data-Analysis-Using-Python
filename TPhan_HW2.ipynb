3. What is the mean number of dependents claimed by all New York tax filers in the adjusted gross income range of $50,000 to $75,000? Display your answer using a meaningful print statement and a formatted numeric answer.

```python
# Calculate the total number of dependents for the rows where agi_stub is 3 (income range of $50,000 to $75,000)
total_dependents = df.loc[df['agi_stub']==3,'NUMDEP'].sum()
```

```python
# display the total number of dependents 
print(total_dependents)
```

```python
# Calculate the total number of returns for the rows where agi_stub is 3 (income range of $50,000 to $75,000)
total_tax_returns = df.loc[df['agi_stub']==3,'N1'].sum()
```

```python
# display the total number of tax returns
print(total_tax_returns)
```

```python
# Calculate the mean number of dependents for the rows where agi_stub is 3 (income range of $50,000 to $75,000)
mean_dependents = total_dependents / total_tax_returns
```

```python
# display the mean number of dependents
print(f'The mean number of dependents claimed by all New York tax filers in the adjusted gross income range of $50,000 to $75,000 is {mean_dependents:.2f}')
```
